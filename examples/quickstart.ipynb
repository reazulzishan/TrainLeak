## 2. A classic mistake on real data
The breast-cancer dataset with a few rows copied, then a random split: the copies end up on both sides.

````
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

df = load_breast_cancer(as_frame=True).frame
df = pd.concat([df, df.sample(40, random_state=0)])   # duplicated rows, as often happens after merging files
train, test = train_test_split(df, test_size=0.2, random_state=42)
trainleak.check(train, test, target="target")
````
