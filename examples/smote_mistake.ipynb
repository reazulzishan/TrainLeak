# A typical thesis notebook with two leaks
Run `trainleak examples/smote_mistake.ipynb` to find them.

```
import pandas as pd
from imblearn.over_sampling import SMOTE
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```
df = pd.read_csv('fraud.csv')
X, y = df.drop(columns='is_fraud'), df['is_fraud']
```

```
scaler = StandardScaler()
X = scaler.fit_transform(X)
X_res, y_res = SMOTE(random_state=42).fit_resample(X, y)   # balance the classes
```

```
X_train, X_test, y_train, y_test = train_test_split(X_res, y_res, test_size=0.2, random_state=42)
model = RandomForestClassifier().fit(X_train, y_train)
print(model.score(X_test, y_test))   # 0.99 ... too good to be true
```
